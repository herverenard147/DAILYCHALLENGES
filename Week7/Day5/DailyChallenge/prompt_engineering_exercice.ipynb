# Exercice — Contrôle du ton, format et longueur en prompt engineering

**Note méthodologique :** ce fichier est un `.ipynb` alors qu'aucune ligne de code Python n'est exécutée ici — tout le contenu est textuel (prompts + sorties + analyses). Si l'exigence de rendu impose ce format, c'est probablement une contrainte de plateforme, pas un besoin réel de notebook exécutable.

## Étape 1 — Prompt : ton, format, longueur

```
Reformule ce texte en français, ton amical et clair, pour des 
employés non techniques.
- Utilise des puces
- Paraphrase, aucune citation directe
- Maximum 75 mots

Texte : "Employees must ensure that all remote access to internal 
systems is established via the approved secure VPN. Under no 
circumstances should unsecured connections or personal devices 
lacking endpoint protection be used to access proprietary data 
or sensitive communications."
```

**Sortie générée :**

> 🔐 **Accès à distance**
> - Connecte-toi toujours via le VPN sécurisé de l'entreprise
> - Évite les connexions non protégées, comme le Wi-Fi public
> - N'utilise jamais un appareil personnel sans protection contre les menaces
> - Les données sensibles doivent rester protégées en toute circonstance

(46 mots)

## Étape 3 — Prompt révisé anti-hallucination

```
Reformule ce texte en français, ton amical et clair, pour des 
employés non techniques.
- Utilise des puces
- Paraphrase, aucune citation directe
- Maximum 75 mots
- N'ajoute aucune information, exemple ou technologie absente 
  du texte source

[texte source]
```

## Étape 4 — Version pour stagiaire junior

```
Explique cette règle à un(e) stagiaire, langage simple, phrases 
courtes, ton bienveillant, sans jargon.
Maximum 4 puces. Base-toi uniquement sur le texte source.

[texte source]
```

**Sortie type :**

> - Connecte-toi toujours au réseau via le VPN de l'entreprise
> - N'utilise jamais ton appareil personnel pour ça
> - Si un appareil n'est pas protégé, ne l'utilise pas
> - Les données sensibles doivent toujours rester protégées

## Étape 5 — Extraction de citation

```
Extrais une seule citation directe du texte, mot pour mot, qui 
résume la règle principale sur l'accès à distance. Pas de commentaire.

[texte source]
```

**Citation extraite :**

> « Employees must ensure that all remote access to internal systems is established via the approved secure VPN. »

**Quand citer plutôt que paraphraser :** documents à portée normative/contractuelle (avenant, audit de conformité, politique officielle) où la formulation exacte fait foi juridiquement.

**Quand citer est risqué :** dans une communication grand public — le jargon reste incompréhensible pour un non-technicien ; et si la politique est mise à jour, une citation figée devient obsolète alors qu'une paraphrase se corrige sans donner l'impression d'avoir mal cité le texte original.

## Angle mort non traité par l'exercice

Aucune des 5 étapes ne vérifie si la reformulation couvre le cas limite : *appareil personnel protégé mais sans VPN*. Le texte source l'interdit clairement (« under no circumstances »), mais toutes les versions simplifiées ci-dessus insistent sur « appareil non protégé = interdit », ce qui peut laisser croire à tort qu'un appareil protégé sans VPN serait acceptable. Un vrai audit de contenu microlearning devrait tester ce cas avant publication.
