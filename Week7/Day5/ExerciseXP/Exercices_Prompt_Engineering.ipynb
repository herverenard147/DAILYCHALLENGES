## Exercice 1 — Post LinkedIn FlowNest

**3 problèmes du prompt vague "Write something about productivity tips" :**
1. Aucun rôle ni voix de marque définis — le modèle ignore qu'il écrit pour FlowNest spécifiquement.
2. Pas de cible ni de plateforme précisées — le ton et les exemples seront génériques.
3. Aucune contrainte de format/longueur — sur LinkedIn un pavé de texte est ignoré.

**Résultat final (le post) :**

> Fatigué de jongler entre 15 onglets et zéro concentration ?
>
> Voici 3 réflexes simples pour retrouver ton focus :
> → Bloque 25 min sans notification (méthode Pomodoro)
> → Une seule tâche à l'écran à la fois
> → Ferme l'appli, pas juste l'onglet
>
> FlowNest t'aide à automatiser ces 3 réflexes, sans discipline de fer.
>
> #Productivité #DeepWork

**Faiblesse que je ne te cache pas** : ce post est interchangeable avec n'importe quelle app de focus concurrente (Forest, Freedom...). Je n'ai aucune fonctionnalité réelle de FlowNest à citer parce que l'énoncé ne m'en fournit aucune. Un post qui ne différencie pas le produit rate l'objectif marketing, même s'il respecte parfaitement le format.


## Exercice 2 — Quiz volcans (7e année)

⚠️ **Aucun article source n'a été fourni.** Ce résumé et ce quiz sont générés à partir de connaissances générales sur le volcanisme, pas d'un texte donné à résumer fidèlement. Si l'exercice est noté sur la fidélité à un article précis, ce résultat est invalide tel quel — il faut me coller le texte réel.

**Résumé (2 puces) :**
- Un volcan entre en éruption quand le magma sous la surface accumule trop de pression et finit par s'échapper à travers une ouverture dans la croûte terrestre.
- La lave, les cendres et les gaz libérés peuvent transformer le paysage, créer de nouvelles terres, mais aussi représenter un danger pour les habitants proches.

**QCM (3 questions) :**

1. Qu'est-ce qui provoque une éruption volcanique ?
   - A) L'accumulation de pression du magma sous la surface ✅
   - B) Un tremblement de terre qui casse la montagne
   - C) La pluie qui s'infiltre et fait bouillir la roche

2. Que produit un volcan lors d'une éruption ?
   - A) De la neige et de la glace
   - B) De la lave, des cendres et des gaz ✅
   - C) Uniquement de la fumée blanche inoffensive

3. Quel effet une éruption peut-elle avoir sur le paysage ?
   - A) Aucun effet visible
   - B) Créer de nouvelles terres ✅
   - C) Faire disparaître définitivement toute vie sur Terre

**Comparaison à la version vague "Make a quiz for kids about this article" :**
1. La vague ne précise pas la tranche d'âge — vocabulaire mal calibré dans un sens ou l'autre.
2. Elle ne précise pas le format de sortie — sans "Google Slides", le contenu rendu est inutilisable tel quel.
3. Elle ne fixe aucune contrainte sur la structure des QCM — nombre de questions et qualité des distracteurs livrés au hasard.

## Exercice 4 — Chatbot support client (style Fonctionnel)

Choix du style **Fonctionnel** : un chatbot support e-commerce doit exécuter une tâche précise (statut de commande) plutôt que brainstormer ou faire la conversation.

**Exemple d'interaction (message client → réponse) :**

> Client : "Ma commande #48213 n'est toujours pas arrivée, elle devait être livrée hier."
>
> Chatbot : "Votre commande #48213 est actuellement en transit, retard estimé de 1 jour dû à un aléa logistique. Livraison prévue demain avant 18h. Voulez-vous que je vous envoie le lien de suivi en temps réel ?"

**2 caractéristiques qui en font un résultat fonctionnel :**
1. Réponse bornée à une tâche précise (statut + action suivante), pas de digression.
2. Contrainte de longueur et une seule action proposée, pas de place à l'improvisation.

**Faiblesse que je dois signaler explicitement** : ce "résultat" est fabriqué — je n'ai aucune donnée réelle de commande #48213. C'est exactement le type de réponse hallucinée que l'exercice 6 te demande d'identifier. Sans connexion à un vrai système de commandes, tout résultat que je te donne ici est une démo, pas une preuve que ça marche en production.

## Exercice 6 — Détection et mitigation de l'hallucination

⚠️ **Aucun article climat/biodiversité marine n'a été fourni.** C'est le seul exercice des six où je ne peux physiquement pas produire un "résumé fidèle" à vérifier sans mentir sur sa fiabilité. Colle l'article et je fais le résumé + vérification claim par claim avec citation des passages sources.

**1. Prompt anti-hallucination (à utiliser une fois l'article fourni) :**
> Résume l'article scientifique fourni ci-dessous sur le changement climatique et la biodiversité marine. N'inclus que des affirmations explicitement présentes dans le texte. Ne complète aucune information à partir de connaissances générales, même si elle te semble plausible ou correcte.

**2. Version avec langage de vérification :**
> Résume l'article fourni. Pour chaque affirmation, cite le passage source correspondant. Si une information n'est pas présente dans l'article, écris "information non disponible dans le texte" plutôt que de la déduire ou de l'estimer.

**3. Stratégies de mitigation :**
- Restriction explicite de la source ("uniquement le texte fourni") pour empêcher le modèle de piocher dans ses connaissances d'entraînement, souvent à l'origine du chiffre halluciné.
- Exiger une citation du passage exact pour chaque affirmation, pas seulement une vérification vague — ça transforme une simple demande de prudence en preuve vérifiable.

**4. Deux domaines à risque élevé :**
- **Juridique** : une clause ou jurisprudence inventée peut orienter une décision légale erronée avec des conséquences contractuelles ou pénales réelles.
- **Médical** : une posologie ou interaction médicamenteuse hallucinée peut mettre en danger la vie d'un patient directement.

**Limite à connaître** : demander une citation exacte réduit le risque mais ne l'élimine pas — un modèle peut halluciner une "citation" qui ne figure pas mot pour mot dans le texte. Une garantie de fiabilité réelle nécessite une vérification humaine ou automatisée de la citation contre le texte original, pas seulement un meilleur prompt.
